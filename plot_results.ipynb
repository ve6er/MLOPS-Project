```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------
# Setup
# -------------------------------
sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.dpi"] = 300
plt.rcParams["savefig.dpi"] = 300

BASE_DIR = "reports_exp"
models = ["lasso", "ridge", "random_forest"]
metrics_list = ["r2_score", "adj_r2_score", "mae", "mse", "rmse"]

OUTPUT_DIR = "model_plots"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------------------
# 1. Read metrics.json summaries
# -------------------------------
summary_data = []
for model in models:
    json_path = os.path.join(BASE_DIR, model, "metrics.json")
    if os.path.exists(json_path):
        with open(json_path, "r") as f:
            metrics = json.load(f)
        metrics["model"] = model
        summary_data.append(metrics)

summary_df = pd.DataFrame(summary_data).set_index("model")
print("✅ Summary metrics loaded:\n", summary_df)

# -------------------------------
# 2. Bar plots for each metric
# -------------------------------
for metric in metrics_list:
    plt.figure(figsize=(7, 5))
    sns.barplot(
        data=summary_df.reset_index(),
        x="model",
        y=metric,
        palette="Set2",
        edgecolor="black"
    )
    plt.title(f"{metric.upper()} Comparison", fontsize=18, weight="bold")
    plt.xlabel("")
    plt.ylabel(metric)
    plt.xticks(rotation=15)
    plt.tight_layout()

    # Save as separate image
    filename = os.path.join(OUTPUT_DIR, f"{metric}_comparison.png")
    plt.savefig(filename, bbox_inches="tight")
    plt.close()
    print(f"💾 Saved: {filename}")

# -------------------------------
# 3. Trend plots (if TSV files exist)
# -------------------------------
trend_data = []
for model in models:
    for metric in metrics_list:
        tsv_path = os.path.join(BASE_DIR, model, "plots", "metrics", f"{metric}.tsv")
        if os.path.exists(tsv_path):
            df = pd.read_csv(tsv_path, sep="\t")
            df["model"] = model
            df["metric"] = metric
            trend_data.append(df)

if trend_data:
    trend_df = pd.concat(trend_data)
    x_col = "step" if "step" in trend_df.columns else trend_df.columns[0]
    y_col = "value" if "value" in trend_df.columns else trend_df.columns[-1]

    for metric in metrics_list:
        plt.figure(figsize=(8, 5))
        subset = trend_df[trend_df["metric"] == metric]
        sns.lineplot(
            data=subset,
            x=x_col,
            y=y_col,
            hue="model",
            palette="Dark2",
            linewidth=2.5,
        )
        plt.title(f"{metric.upper()} Trend Across Models", fontsize=18, weight="bold")
        plt.xlabel("Training Step")
        plt.ylabel(metric)
        plt.legend(title="Model", loc="best")
        plt.tight_layout()

        filename = os.path.join(OUTPUT_DIR, f"{metric}_trend.png")
        plt.savefig(filename, bbox_inches="tight")
        plt.close()
        print(f"💾 Saved: {filename}")

# -------------------------------
# 4. Correlation heatmap
# -------------------------------
plt.figure(figsize=(6, 5))
sns.heatmap(summary_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Metrics", fontsize=16, weight="bold")
plt.tight_layout()

filename = os.path.join(OUTPUT_DIR, "metric_correlation_heatmap.png")
plt.savefig(filename, bbox_inches="tight")
plt.close()
print(f"💾 Saved: {filename}")

print("\n🎉 All plots saved successfully in:", OUTPUT_DIR)
```
